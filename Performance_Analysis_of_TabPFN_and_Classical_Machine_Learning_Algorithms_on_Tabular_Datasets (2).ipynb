## 2. Abstract

Tabular data is widely used in real-world applications such as healthcare, finance, business, and scientific research. Classical machine learning algorithms, including Logistic Regression, Decision Trees, Random Forest, Support Vector Machines, and Linear Regression, are commonly applied to such datasets. Recently, TabPFN (Tabular Prior-Data Fitted Network) has emerged as a transformer-based approach specifically designed for tabular data and has demonstrated strong performance with limited model training and preprocessing requirements.

This project presents a systematic performance analysis of TabPFN against selected classical machine learning algorithms on multiple tabular datasets. The study considers both classification and regression tasks to evaluate the models under different data characteristics. The datasets are analysed using descriptive statistics, exploratory data analysis, data preprocessing, and appropriate statistical tests before model development. TabPFN and classical machine learning models are trained using controlled experimental conditions and evaluated using task-specific performance measures such as accuracy, precision, recall, F1-score, confusion matrix, MAE, RMSE, and R². Computational aspects such as training and prediction time are also considered where applicable. The experimental results are used to determine whether TabPFN provides consistent performance advantages over classical machine learning algorithms across different types of tabular datasets.


## 3. Introduction

Tabular data is one of the most common forms of structured data used in machine learning. It consists of observations represented as rows and features represented as columns, with a target variable that can be used for prediction or classification. Many practical machine learning problems, including disease diagnosis, income prediction, customer analysis, and housing-price prediction, are based on tabular datasets.

Classical machine learning algorithms have traditionally been the primary methods for solving tabular-data problems. Algorithms such as Logistic Regression, Linear Regression, Decision Trees, Random Forests, K-Nearest Neighbors, Support Vector Machines, and Gradient Boosting can provide effective results when appropriate preprocessing, feature engineering, and parameter selection are performed. However, their performance can vary depending on the size, structure, feature types, and complexity of the dataset.

Recent developments in deep learning have introduced specialised architectures for tabular data. TabPFN (Tabular Prior-Data Fitted Network) is a transformer-based model designed specifically for tabular prediction tasks. Instead of following the conventional approach of training a model from scratch for every dataset, TabPFN uses prior knowledge obtained during pretraining to make predictions on new tabular datasets. This provides an alternative approach to conventional machine learning methods.

The emergence of TabPFN raises an important question: how does a pretrained tabular model perform compared with established classical machine learning algorithms? A comparison based on only one dataset or one performance metric may not provide a reliable conclusion. Therefore, this project evaluates the models across multiple tabular datasets containing different characteristics and considers both classification and regression tasks.

The study follows a systematic experimental procedure involving data understanding, descriptive statistics, exploratory data analysis, preprocessing, model training, performance evaluation, and statistical comparison. The objective is not only to identify the model with the highest performance but also to understand the conditions under which TabPFN and classical machine learning algorithms perform effectively.

The Breast Cancer Wisconsin Diagnostic (WDBC) dataset is a binary classification dataset containing measurements computed from digitised images of breast-mass cell nuclei. It contains **569 observations and 30 numerical features**. The target represents whether the tumour is malignant or benign.

This dataset is selected because its relatively small size makes it suitable for evaluating how TabPFN performs on small tabular classification problems.

| Property | Value |
|---|---|

| Target | `diagnosis` (0 = Malignant, 1 = Benign) |
| Features | 30 numerical |
| Samples | 569 |

**Feature groups.** The 30 features are ten base morphological measurements of the cell nuclei — radius, texture, perimeter, area, smoothness, compactness, concavity, concave points, symmetry and fractal dimension — each reported in three variants: **mean**, **error** (standard error) and **worst** (mean of the three largest values). Because the three variants describe the same underlying quantities, substantial correlation between features is expected; this is examined directly in the correlation analysis rather than assumed.

## 9. Data Quality Analysis

Before performing EDA and training the models, we determine whether the datasets contain missing values, duplicate records, incorrect data types, or problematic categorical values.

This matters for two distinct reasons:

1. **Model performance.** Undetected missing values and mis-typed columns propagate silently into the modelling stage and degrade results in ways that are hard to diagnose later.
2. **Fairness of the comparison.** If the datasets are not handled consistently across models, the TabPFN vs. classical ML comparison measures preprocessing differences rather than algorithmic differences. Data-quality decisions are therefore made once, at the dataset level, and applied identically to every model.


### 9.1 Missing Value Analysis

Missing values are counted per column for each dataset. Beyond the raw count, what matters is *which* columns are affected and *how* the missingness is encoded — a value recorded as `?` will not be counted by `isnull()` until it is converted.


````python
# ============================================================
# 9.1 MISSING VALUE ANALYSIS
# ============================================================

print("=" * 70)
print("MISSING VALUES - BREAST CANCER")
print("=" * 70)

print(breast_cancer_df.isnull().sum())


print("\n" + "=" * 70)
print("MISSING VALUES - ADULT INCOME")
print("=" * 70)

print(adult_df.isnull().sum())


print("\n" + "=" * 70)
print("MISSING VALUES - CALIFORNIA HOUSING")
print("=" * 70)

print(california_housing_df.isnull().sum())
````

````python
# ============================================================
# MISSING VALUE SUMMARY
# ============================================================

def missing_value_summary(df, dataset_name):
    summary = pd.DataFrame({
        "Missing Count": df.isnull().sum(),
        "Missing Percentage": (df.isnull().mean() * 100).round(2)
    })

    summary = summary[summary["Missing Count"] > 0]

    print(f"\n{dataset_name}")

    if summary.empty:
        print("No missing values found.")
    else:
        display(summary)


missing_value_summary(
    breast_cancer_df,
    "Breast Cancer Dataset"
)

missing_value_summary(
    adult_df,
    "Adult Income Dataset"
)

missing_value_summary(
    california_housing_df,
    "California Housing Dataset"
)
````

#### Missing Value Results

1. **Breast Cancer:** **no missing values** in any of the 31 columns.
2. **California Housing:** **no missing values** in any of the 9 columns.
3. **Adult Income:** missing values in **three columns** — `workclass` (1,836 rows, 5.64%), `occupation` (1,843 rows, 5.66%) and `native_country` (583 rows, 1.79%).
4. **Pattern worth noting:** `workclass` and `occupation` are missing at almost identical rates, which suggests the two may be missing together for the same records rather than independently — plausible if both are unrecorded for individuals outside the workforce.
5. **Decision:** the affected proportions are small (under 6%), so these rows are retained and the values are imputed inside the training pipeline rather than dropped. Dropping them would discard roughly 7% of the dataset and risk introducing bias if the missingness is not random.

Duplicate rows are checked because they distort both training and evaluation: an identical record appearing in both the training and test sets inflates the apparent performance, since the model has effectively seen the test observation already.


````python
# ============================================================
# 9.2 DUPLICATE RECORD ANALYSIS
# ============================================================

datasets = {
    "Breast Cancer": breast_cancer_df,
    "Adult Income": adult_df,
    "California Housing": california_housing_df
}

duplicate_summary = []

for name, df in datasets.items():
    duplicate_summary.append({
        "Dataset": name,
        "Duplicate Rows": df.duplicated().sum(),
        "Total Rows": len(df)
    })

duplicate_df = pd.DataFrame(duplicate_summary)

display(duplicate_df)
````

### 9.4 Feature Type Identification

Feature types are identified programmatically rather than hard-coded, so that the correct transformer can be applied to each column group in the preprocessing pipeline.


````python
# ============================================================
# 9.4 FEATURE TYPE IDENTIFICATION
# ============================================================

def feature_type_summary(df, target_column):

    X = df.drop(columns=[target_column])

    numerical_features = X.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_features = X.select_dtypes(
        exclude=np.number
    ).columns.tolist()

    print("Numerical Features:")
    print(numerical_features)

    print("\nNumber of Numerical Features:", len(numerical_features))

    print("\nCategorical Features:")
    print(categorical_features)

    print("\nNumber of Categorical Features:", len(categorical_features))


print("=" * 70)
print("BREAST CANCER")
print("=" * 70)

feature_type_summary(
    breast_cancer_df,
    "diagnosis"
)


print("\n" + "=" * 70)
print("ADULT INCOME")
print("=" * 70)

feature_type_summary(
    adult_df,
    "income"
)


print("\n" + "=" * 70)
print("CALIFORNIA HOUSING")
print("=" * 70)

feature_type_summary(
    california_housing_df,
    "median_house_value"
)
````

### 9.6 Class Distribution — Classification Datasets

Class imbalance matters because accuracy alone becomes misleading when classes are not evenly represented. If one class accounts for 76% of observations, a model that always predicts that class already scores 76% accuracy while being useless. This is why the evaluation protocol in Section 15 reports precision, recall, F1 and ROC-AUC alongside accuracy.


````python
# ============================================================
# 9.6 CLASS DISTRIBUTION
# ============================================================

print("BREAST CANCER CLASS DISTRIBUTION")
print(
    breast_cancer_df["diagnosis"].value_counts()
)

print("\nClass Distribution (%)")
print(
    breast_cancer_df["diagnosis"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
````

````python
print("ADULT INCOME CLASS DISTRIBUTION")
print(
    adult_df["income"].value_counts()
)

print("\nClass Distribution (%)")
print(
    adult_df["income"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
````

#### Class Distribution Results

1. **Breast Cancer:** 357 Benign (**62.74%**) and 212 Malignant (**37.26%**) — a **mild imbalance**. No resampling is needed, but accuracy alone would still be misleading, since always predicting Benign already yields 62.74%.
2. **Adult Income:** 24,720 `<=50K` (**75.92%**) and 7,841 `>50K` (**24.08%**) — a **moderate imbalance**, roughly 3:1.

## 10. Exploratory Data Analysis (EDA)

EDA is performed separately for each dataset. The purpose is to understand the distribution, spread and relationships of the variables *before* modelling, so that preprocessing decisions are informed by the data rather than applied by default.

### 10.1 Descriptive Statistics

#### 10.1.1 Mean, Median and Mode

**Why:** these measures describe the *central tendency* of numerical variables — where the bulk of the data sits.

| Measure | Definition | What it tells us |
|---|---|---|
| **Mean** | Arithmetic average | Sensitive to outliers; shifts toward extreme values |
| **Median** | Middle value when sorted | Robust to outliers |
| **Mode** | Most frequently occurring value | Useful for discrete or categorical-like variables |

A large gap between the mean and the median is itself informative: it indicates skewness, and therefore that symmetric summaries (such as mean ± standard deviation) may describe the variable poorly.


#### Computation


````python
# ============================================================
# 10.1.1 MEAN, MEDIAN AND MODE
# ============================================================

def descriptive_central_tendency(df, dataset_name):
    numerical_df = df.select_dtypes(include=np.number)

    summary = pd.DataFrame({
        "Mean": numerical_df.mean(),
        "Median": numerical_df.median(),
        "Mode": numerical_df.mode().iloc[0]
    })

    print(f"\n{dataset_name}")
    display(summary)


descriptive_central_tendency(
    breast_cancer_df,
    "Breast Cancer Dataset"
)

descriptive_central_tendency(
    adult_df,
    "Adult Income Dataset"
)

descriptive_central_tendency(
    california_housing_df,
    "California Housing Dataset"
)
````

#### 10.2 Variance and Standard Deviation


````python
# ============================================================
# 10.2 VARIANCE AND STANDARD DEVIATION
# ============================================================

def variability_analysis(df, dataset_name):
    numerical_df = df.select_dtypes(include=np.number)

    summary = pd.DataFrame({
        "Variance": numerical_df.var(),
        "Standard Deviation": numerical_df.std()
    })

    print(f"\n{dataset_name}")
    display(summary)


variability_analysis(
    breast_cancer_df,
    "Breast Cancer Dataset"
)

variability_analysis(
    adult_df,
    "Adult Income Dataset"
)

variability_analysis(
    california_housing_df,
    "California Housing Dataset"
)
````

#### 10.2 Interpretation — Variance and Standard Deviation

**Variance** measures how spread out the observations are around the mean. It is expressed in squared units, which makes it hard to interpret directly.

**Standard deviation** is the square root of the variance and therefore shares the units of the original variable, representing the typical amount of variation from the mean.

**Why this matters for this project:** features with very different standard deviations occupy very different numeric ranges. Distance-based and margin-based algorithms (KNN, SVM) and regularised linear models are directly affected by this, because a feature with a large range dominates the distance calculation regardless of its predictive relevance. This is precisely why scaling is applied to those models in Section 11 — and why tree-based models, which split on thresholds within each feature independently, are largely unaffected.


#### Computation — Quartiles and IQR


````python
# ============================================================
# 10.3 QUARTILES AND INTERQUARTILE RANGE
# ============================================================

def quartile_analysis(df, dataset_name):
    numerical_df = df.select_dtypes(include=np.number)

    Q1 = numerical_df.quantile(0.25)
    Q2 = numerical_df.quantile(0.50)
    Q3 = numerical_df.quantile(0.75)

    IQR = Q3 - Q1

    summary = pd.DataFrame({
        "Q1 (25%)": Q1,
        "Median (50%)": Q2,
        "Q3 (75%)": Q3,
        "IQR": IQR
    })

    print(f"\n{dataset_name}")
    display(summary)


quartile_analysis(
    breast_cancer_df,
    "Breast Cancer Dataset"
)

quartile_analysis(
    adult_df,
    "Adult Income Dataset"
)

quartile_analysis(
    california_housing_df,
    "California Housing Dataset"
)
````

#### 10.3 Quartiles and the Interquartile Range

Quartiles divide the sorted data into four equal parts: **Q1** (25th percentile), **Q2** (the median, 50th percentile) and **Q3** (75th percentile).

The Interquartile Range is:

$$ IQR = Q_3 - Q_1 $$

It represents the spread of the middle 50% of observations, and — unlike the standard deviation — is **robust to outliers**, because it ignores the extreme quarters of the data entirely.

The IQR also provides the standard rule-of-thumb for outlier detection used in the boxplots below: an observation is flagged as a potential outlier if it lies below $Q_1 - 1.5 \times IQR$ or above $Q_3 + 1.5 \times IQR$.

**Purpose:** histograms show the *shape* of a variable's distribution, which summary statistics alone cannot convey. Two variables can share a mean and standard deviation while having completely different distributions.

They allow us to inspect:

**Why this matters here:** strongly skewed features affect linear models more than tree-based models, and heavy tails influence how much scaling changes model behaviour.

````python
# ============================================================
# 10.6 CORRELATION MATRIX
# ============================================================

def correlation_analysis(df, dataset_name):

    numerical_df = df.select_dtypes(
        include=np.number
    )

    correlation_matrix = numerical_df.corr()

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        correlation_matrix,
        annot=False,
        cmap="coolwarm",
        center=0
    )

    plt.title(f"{dataset_name} - Correlation Matrix")
    plt.tight_layout()
    plt.show()


correlation_analysis(
    breast_cancer_df,
    "Breast Cancer"
)

correlation_analysis(
    adult_df,
    "Adult Income"
)

correlation_analysis(
    california_housing_df,
    "California Housing"
)
````

#### Correlation Matrix Observations

1. **Strong block structure is visible** in the Breast Cancer heatmap. The mean, error and worst variants of the same base measurement form clearly correlated groups.
2. **Geometrically related features are near-perfectly correlated** — radius, perimeter and area measure the same underlying object size and are related by definition, not merely by empirical association.
3. **Multicollinearity is therefore present and expected.** This is a property of how the features were constructed rather than a data-quality problem.
4. **Consequence for interpretation, not for accuracy.** The correlated features will destabilise Logistic Regression's *coefficient* estimates, so those coefficients should not be read as individual feature effects. Predictive accuracy is largely unaffected, and tree-based models and TabPFN handle the redundancy without difficulty.
5. **No features are removed on the basis of correlation.** Dropping correlated features would be a modelling decision that should be validated rather than assumed, and the models used here do not require it.


#### 10.6 Interpretation — Correlation Analysis

**Important:** correlation analysis is *exploratory*, and correlation does **not** establish causation. A high correlation between two variables may arise from a direct relationship, a shared underlying cause, or coincidence in a finite sample.

What correlation analysis is used for here:

1. **Understanding feature relationships** — which variables carry related information.
2. **Detecting redundancy (multicollinearity)** — in the Breast Cancer dataset, the mean/error/worst variants of the same morphological measurement are expected to be strongly correlated, since they describe the same underlying quantity.

**Why multicollinearity matters selectively.** Highly correlated predictors destabilise the *coefficients* of linear models — the coefficient estimates become sensitive to small changes in the data, which makes them unreliable for interpretation even when predictive accuracy is unaffected. Tree-based ensembles and TabPFN are far less affected in terms of predictive performance, although correlated features do split importance between themselves, which complicates feature-importance interpretation.


## 11. Data Preprocessing

This section matters because TabPFN and classical ML models should not necessarily receive identical preprocessing. Classical algorithms require numerical input, so categorical variables must be encoded and scale-sensitive models require standardisation. TabPFN is designed to consume tabular data in a closer-to-raw form.

### 11.0 Preprocessing Strategy

| Dataset | TabPFN | Classical ML |
|---|---|---|
| Breast Cancer | Raw numerical features | Scaling where required |
| Adult Income | Numerical + categorical features | Imputation + encoding + scaling where required |
| California Housing | Raw numerical features | Scaling where required |

**Which models need scaling, and why:**

| Requires scaling | Reason |
|---|---|
| Logistic Regression | Regularisation penalises large coefficients; unscaled features distort the penalty |
| KNN | Distance is dominated by large-range features |
| SVM | The margin and the RBF kernel both depend on distances |

| Generally does not require scaling | Reason |
|---|---|
| Decision Tree | Splits on thresholds within each feature independently |
| Random Forest | Ensemble of such trees |
| Gradient Boosting | Ensemble of such trees |

**Data leakage control.** All preprocessing parameters — imputation statistics, scaling means and standard deviations, encoder categories — are learned **only from the training data** and then applied to the test data. `ColumnTransformer` and `Pipeline` are used so that this is enforced structurally rather than by manual discipline. The target variable is never scaled.

### 11.5 Prepare Classical ML Preprocessing

For classical ML models, the Adult dataset requires:

1. **Missing-value imputation** — the `?` values converted to `NaN` must be filled
2. **Categorical encoding** — one-hot encoding, since classical algorithms require numerical input
3. **Numerical scaling** — for Logistic Regression, KNN and SVM

`ColumnTransformer` and `Pipeline` are used so that every one of these steps is fitted **only on the training data** and then applied to the test data. This is the structural guarantee against data leakage: because the transformations live inside the pipeline, it is not possible to accidentally fit them on the full dataset.

#### 11.5.1 Numerical Preprocessing


````python
# ============================================================
# 11.5 CLASSICAL ML PREPROCESSING
# ============================================================

from sklearn.impute import SimpleImputer

# Numerical preprocessing
numeric_preprocessor = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
````

#### 11.6 Categorical Preprocessing

Missing categorical values are imputed before encoding, because the one-hot encoder cannot process nulls directly.


````python
# Categorical preprocessing
categorical_preprocessor = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)
````

### 11.7 Adult Income Preprocessor

The numerical and categorical transformers are combined into a single `ColumnTransformer`, which applies each to the correct subset of columns in one step.


````python
# ============================================================
# ADULT INCOME PREPROCESSOR
# ============================================================

adult_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_preprocessor,
            adult_numeric_features
        ),
        (
            "cat",
            categorical_preprocessor,
            adult_categorical_features
        )
    ]
)

print("Adult Income preprocessing pipeline created successfully.")
````

### 11.8 Breast Cancer and California Housing Preprocessors

These datasets contain only numerical features, so their preprocessors require scaling alone — no imputation and no encoding.


````python
# ============================================================
# BREAST CANCER PREPROCESSOR
# ============================================================

breast_preprocessor = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

print("Breast Cancer preprocessing pipeline created successfully.")

# ============================================================
# CALIFORNIA HOUSING PREPROCESSOR
# ============================================================

housing_preprocessor = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

print("California Housing preprocessing pipeline created successfully.")
````

## 13. TabPFN Model

TabPFN (Tabular Prior-Data Fitted Network) is a transformer pre-trained once on a large collection of *synthetic* tabular prediction tasks. Rather than fitting parameters to each new dataset by iterative optimisation, it performs **in-context learning**: the labelled training rows and the unlabelled test rows are supplied together, and the model approximates the posterior predictive distribution for the test rows conditioned on the training rows.

**Practical consequence for this project:** `fit()` is close to a data-ingestion step, and the substantive computation happens during **prediction** — the inverse of the classical cost profile, where training is expensive and inference is nearly free. This is directly visible in the computational results in Section 18, and it is the reason training time alone would be a misleading efficiency measure here.

**Preprocessing requirements:** TabPFN does not require feature scaling, and handles the tabular structure directly, which is why it receives the raw-feature path defined in Section 11.9.

````python
# ============================================================
# 14.4 CLASSICAL ML MODELS - BREAST CANCER
# ============================================================

print("=" * 70)
print("CLASSICAL ML MODELS - BREAST CANCER")
print("=" * 70)

# ------------------------------------------------------------
# Define models
# ------------------------------------------------------------

breast_models = {

    "Logistic Regression": Pipeline([
        ("preprocessor", breast_preprocessor),
        ("model", LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ]),

    "KNN": Pipeline([
        ("preprocessor", breast_preprocessor),
        ("model", KNeighborsClassifier(
            n_neighbors=5
        ))
    ]),

    "Decision Tree": Pipeline([
        ("preprocessor", Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ])),
        ("model", DecisionTreeClassifier(
            random_state=42
        ))
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ])),
        ("model", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ))
    ]),

    "SVM": Pipeline([
        ("preprocessor", breast_preprocessor),
        ("model", SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ))
    ])
}


# ------------------------------------------------------------
# Train and predict
# ------------------------------------------------------------

breast_classical_results = {}
breast_classical_predictions = {}

for model_name, model in breast_models.items():

    print(f"\nTraining {model_name}...")

    # Training time
    start_time = time.time()

    model.fit(
        X_breast_classical_train,
        y_breast_classical_train
    )

    training_time = time.time() - start_time

    # Prediction time
    start_time = time.time()

    y_pred = model.predict(
        X_breast_classical_test
    )

    y_prob = model.predict_proba(
        X_breast_classical_test
    )

    prediction_time = time.time() - start_time

    # Store model
    breast_classical_predictions[model_name] = {
        "y_true": y_breast_classical_test.copy(),
        "y_pred": y_pred.copy(),
        "y_prob": y_prob.copy()
    }

    # Store timing
    breast_classical_results[model_name] = {
        "Training Time (s)": training_time,
        "Prediction Time (s)": prediction_time
    }

    print(f"Training time   : {training_time:.4f} seconds")
    print(f"Prediction time : {prediction_time:.4f} seconds")


print("\n" + "=" * 70)
print("ALL BREAST CANCER CLASSICAL MODELS COMPLETED")
print("=" * 70)
````

## 15. Model Evaluation

All models are evaluated using the predictions generated in the previous sections, on the same test observations.

### 15.0 Metrics Used

**Classification metrics**

| Metric | Definition | Interpretation |
|---|---|---|
| **Accuracy** | $(TP+TN)/(TP+TN+FP+FN)$ | Overall proportion correct; misleading under imbalance |
| **Precision** | $TP/(TP+FP)$ | Of those predicted positive, how many are correct |
| **Recall** | $TP/(TP+FN)$ | Of the actual positives, how many were found |
| **F1 Score** | $2 \cdot \frac{P \cdot R}{P + R}$ | Harmonic mean of precision and recall |
| **ROC-AUC** | Area under the ROC curve | Ranking quality across all thresholds |

**Regression metrics**

| Metric | Interpretation |
|---|---|
| **MAE** | Mean absolute error — average error size, in target units |
| **RMSE** | Root mean squared error — penalises large errors more heavily |
| **R²** | Proportion of target variance explained by the model |

**Why not accuracy alone.** On Adult Income, 75.92% of observations belong to the `<=50K` class, so a model predicting that class for every record would score 75.92% accuracy while identifying no high-income individuals at all. F1 and ROC-AUC are therefore treated as the primary classification evidence.

**Note:** Adult Income and California Housing use the matched 2,000-training / 1,000-test controlled experiment for both TabPFN and the classical models.


### 15.1 Evaluation Functions

Two reusable evaluation functions are defined — one for classification and one for regression — so that every model is scored by identical code. This removes the possibility of a model being advantaged by a subtly different metric computation.


````python
# ============================================================
# 15.1 EVALUATION FUNCTIONS
# ============================================================

def evaluate_classification_model(dataset_name, model_name, y_true, y_pred, y_prob):
    return {
        "Dataset": dataset_name,
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1 Score": f1_score(y_true, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_true, y_prob[:, 1])
    }


def evaluate_regression_model(dataset_name, model_name, y_true, y_pred):
    return {
        "Dataset": dataset_name,
        "Model": model_name,
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R²": r2_score(y_true, y_pred)
    }
````

## 17. Statistical Analysis

A difference in a performance metric is not by itself evidence that one model is better — the difference may fall within the range expected from sampling variation alone. Statistical tests are therefore used to determine whether the observed differences are larger than chance would explain.

### 17.0 Test Design

Statistical tests compare TabPFN with each classical model using **paired predictions from the same test observations**. Pairing is what makes the tests appropriate here: because every model predicted the same rows, the comparison controls for the difficulty of individual observations.

| Task | Test | Why this test |
|---|---|---|
| Classification | Exact McNemar test | Designed for paired binary outcomes; tests whether disagreements between two models are asymmetric |
| Regression | Wilcoxon signed-rank test on absolute errors | Non-parametric paired test; does not assume normally distributed errors |

**Hypotheses (McNemar):**
- **H₀:** the two models have the same error rate — disagreements are equally likely in both directions.
- **H₁:** the two models have different error rates.

**Hypotheses (Wilcoxon):**
- **H₀:** the median difference in absolute error between the two models is zero.
- **H₁:** the median difference is non-zero.

**Significance level:** α = 0.05.

**Multiple-comparison correction.** Five pairwise comparisons are made within each classification dataset. Testing repeatedly at α = 0.05 inflates the probability of at least one false positive, so the **Holm correction** is applied and both raw and adjusted p-values are reported.

**Limitation.** These tests assess whether differences on *this particular test set* are unlikely under the null hypothesis. They do not establish that the advantage would generalise to other datasets, other splits, or other samples, and statistical significance does not imply that a difference is large enough to matter practically.


### 17.1 Statistical Test Functions


````python
# ============================================================
# 17.1 STATISTICAL TEST FUNCTIONS
# ============================================================

from scipy.stats import binomtest, wilcoxon


def mcnemar_exact_test(y_true, pred_model_1, pred_model_2):
    y_true = np.asarray(y_true)
    pred_model_1 = np.asarray(pred_model_1)
    pred_model_2 = np.asarray(pred_model_2)

    correct_1 = pred_model_1 == y_true
    correct_2 = pred_model_2 == y_true

    b = int(np.sum(correct_1 & ~correct_2))
    c = int(np.sum(~correct_1 & correct_2))
    n_discordant = b + c

    if n_discordant == 0:
        statistic = 0
        p_value = 1.0
    else:
        statistic = min(b, c)
        p_value = binomtest(
            min(b, c),
            n=n_discordant,
            p=0.5,
            alternative="two-sided"
        ).pvalue

    return statistic, p_value


def wilcoxon_regression_test(y_true, pred_model_1, pred_model_2):
    y_true = np.asarray(y_true)
    pred_model_1 = np.asarray(pred_model_1)
    pred_model_2 = np.asarray(pred_model_2)

    error_1 = np.abs(y_true - pred_model_1)
    error_2 = np.abs(y_true - pred_model_2)

    result = wilcoxon(
        error_1,
        error_2,
        zero_method="wilcox",
        alternative="two-sided"
    )

    return result.statistic, result.pvalue


def holm_adjust(p_values):
    p_values = np.asarray(p_values, dtype=float)
    m = len(p_values)

    order = np.argsort(p_values)
    adjusted = np.empty(m)
    running_max = 0.0

    for rank, idx in enumerate(order):
        value = (m - rank) * p_values[idx]
        running_max = max(running_max, value)
        adjusted[idx] = min(running_max, 1.0)

    return adjusted
````

### 17.2 Breast Cancer Statistical Analysis


````python
# ============================================================
# 17.2 BREAST CANCER STATISTICAL ANALYSIS
# ============================================================

rows = []

for model_name, prediction_data in breast_classical_predictions.items():
    statistic, p_value = mcnemar_exact_test(
        y_breast_test,
        y_breast_pred_tabpfn,
        prediction_data["y_pred"]
    )

    rows.append({
        "Dataset": "Breast Cancer",
        "Model 1": "TabPFN",
        "Model 2": model_name,
        "Test": "McNemar Exact",
        "Statistic": statistic,
        "p-value": p_value
    })

breast_statistical_df = pd.DataFrame(rows)
breast_statistical_df["Holm Adjusted p-value"] = holm_adjust(
    breast_statistical_df["p-value"].values
)
breast_statistical_df["Result"] = np.where(
    breast_statistical_df["Holm Adjusted p-value"] < 0.05,
    "Significant",
    "Not Significant"
)

display(breast_statistical_df.round(4))
````

#### Breast Cancer Statistical Results

1. **No comparison is statistically significant after Holm correction.** All five adjusted p-values are 1.0000 or above the 0.05 threshold.
2. **The Decision Tree comparison is the critical case.** Its raw p-value of **0.0215** is below α = 0.05 and would have been declared significant — but the Holm-adjusted value rises to **0.1074**, which is not significant.
3. **This is exactly why the correction is applied.** Reporting raw p-values alone would have produced a claim that five simultaneous comparisons do not support.
4. **McNemar statistics of 0 and 1** against Logistic Regression and SVM mean the models disagreed on at most one or two test observations in total — they are making effectively the same predictions.

## 20. Discussion

The results are interpreted jointly using predictive performance, statistical significance and computational cost, because a model that wins on one of these three can lose decisively on another.

### 20.1 Predictive Performance

For classification, F1 Score provides the primary comparison while Accuracy, Precision, Recall and ROC-AUC provide additional evidence. For regression, R² is the primary comparison while MAE and RMSE describe prediction error in the target's own units.

**Breast Cancer.** TabPFN reached an F1 of 0.9861 — but Logistic Regression and SVM reached exactly the same F1, with only a marginal TabPFN advantage in ROC-AUC (0.9967 vs. 0.9954 and 0.9950). *A possible explanation:* this dataset appears to be close to linearly separable in its 30-dimensional feature space, so once a model captures that structure, there is very little headroom left for a more expressive method to exploit. Where the ceiling is this high, model choice matters less than it does on harder problems.

**Adult Income.** TabPFN achieved the highest F1 (0.6712) with Random Forest closest at 0.6590 — a difference of 0.0122. The absolute F1 values are much lower than on Breast Cancer for all models, which reflects the genuine difficulty of the task rather than a modelling failure: income is only partially determined by the recorded demographic attributes.

**California Housing.** This is where the largest gap appears. TabPFN reached R² = 0.8561 against 0.7575 for the best classical model, Gradient Boosting — a difference of 0.0985. *A possible explanation:* the relationship between the housing features and median value is strongly non-linear and involves geographic interaction effects; Linear Regression's R² of 0.5014 is consistent with this reading, and the ensembles close much but not all of the gap.

### 20.2 Statistical Evidence

The pattern of significance is as informative as the metric values themselves.

On **Breast Cancer**, *no* comparison remained significant after Holm correction — including the Decision Tree comparison, whose raw p-value of 0.0215 rose to 0.1074 once corrected. The apparent TabPFN advantage on this dataset is therefore **not statistically supported**.

On **Adult Income**, only the comparisons against KNN (adjusted p = 0.0122) and Decision Tree (adjusted p = 0.0219) were significant. The comparisons against Logistic Regression, Random Forest and SVM were not — so TabPFN's F1 advantage over the strongest classical baselines on this dataset is also **not statistically supported**.

On **California Housing**, all four comparisons were significant. This is the one dataset where the TabPFN advantage is both large and statistically supported.

**An important caution:** the Holm correction changed the conclusion on Breast Cancer. Reporting only raw p-values would have produced a claim the data does not support, which is why the correction is applied rather than omitted.

### 20.3 Computational Cost

The cost profiles are inverted exactly as expected. On Breast Cancer, TabPFN required 74.18 s to fit and 18.29 s to predict, against 0.0575 s and 0.0066 s for Logistic Regression — a total-time ratio of roughly **1,400×** for identical F1. On California Housing, TabPFN's 147.83 s total compares with 1.31 s for Gradient Boosting.

This is the clearest practical finding in the study: TabPFN's accuracy advantage, where it exists, is purchased at a computational cost two to three orders of magnitude higher in this CPU-only setting. Whether that trade is worthwhile depends entirely on the application — negligible for a one-off analysis of 569 records, potentially prohibitive for high-throughput inference.

### 20.4 Scope of the Conclusions

The Adult Income and California Housing experiments are **controlled subset experiments** (2,000 training / 1,000 test) necessitated by the CPU-only execution environment. They should not be described as full-dataset TabPFN results. All conclusions are based on the numerical outputs produced above rather than on an assumption that TabPFN is universally superior.

## 21. Conclusion

This study systematically compared TabPFN with classical machine learning algorithms across two binary classification datasets and one regression dataset, evaluating them on predictive performance, statistical significance and computational cost.

### 21.1 Answers to the Research Questions

**RQ1 — Does TabPFN achieve better predictive performance?** On the metric values, yes on all three datasets. On the statistical evidence, only on California Housing. TabPFN's numerical advantage on Breast Cancer and over the strongest classical baselines on Adult Income did not survive Holm correction.

**RQ2 — Does performance vary across task types?** Yes, and this was the strongest pattern observed. TabPFN's largest and only statistically supported advantage was on the **regression** task (ΔR² = 0.0985), while on classification it tied or led marginally without statistical support.

**RQ3 — Is the advantage consistent across dataset sizes and characteristics?** No. It was absent on the smallest dataset, marginal on the mixed-type dataset, and substantial on the numerical regression dataset.

**RQ4 — How does TabPFN handle mixed numerical and categorical features?** It achieved the highest F1 on Adult Income (0.6712), but its margin over Random Forest (0.6590) was not statistically significant.

**RQ5 — Are the differences statistically significant?** Only in 6 of the 14 pairwise comparisons after Holm correction — all four on California Housing, and two on Adult Income (against KNN and Decision Tree).

**RQ6 — How does computational efficiency compare?** Classical models were faster by two to three orders of magnitude in total time. TabPFN's cost is concentrated in prediction rather than training, inverting the classical profile.

**RQ7 — Which classical algorithm was most competitive?** Logistic Regression on Breast Cancer (matching TabPFN's F1 exactly), Random Forest on Adult Income, and Gradient Boosting on California Housing.

### 21.2 Overall Conclusion

The results demonstrate that **no single algorithm is consistently superior across all tabular learning problems.** TabPFN is highly competitive and can achieve strong predictive performance on small and medium-sized tabular datasets, particularly when minimal preprocessing is required. However, classical models such as Random Forest, SVM and Gradient Boosting remained highly competitive, and on Breast Cancer, Logistic Regression matched TabPFN exactly.

The main advantage of TabPFN is its ability to provide strong predictions with minimal feature engineering and limited hyperparameter tuning, which is attractive when rapid experimentation and reduced preprocessing effort matter. Classical algorithms offer greater flexibility, scalability, interpretability and computational practicality, especially for larger datasets.

**TabPFN should not be considered a universal replacement for classical machine learning.** It is better understood as a powerful alternative that can be particularly effective on suitable tabular datasets — in this study, most clearly on the non-linear regression problem. Classical ensemble methods remain strong baselines. The better model depends on the dataset, the task type, the evaluation metric, the computational resources available and the practical requirements of the application.

### 21.3 Limitations

1. **Number of datasets.** Three datasets cannot establish general conclusions about tabular learning; they can only characterise behaviour on these problems.
2. **Subset experiments.** The Adult Income and California Housing results use 2,000-training / 1,000-test subsets and are not full-dataset benchmarks.
3. **CPU-only environment.** All timings reflect CPU execution. TabPFN is designed for GPU inference, so the computational gap reported here is an upper bound rather than an intrinsic property.
4. **Single train/test split.** Performance is estimated from one split per dataset rather than from repeated cross-validation, so the estimates carry more variance than a cross-validated result would.
5. **Limited hyperparameter tuning.** The classical models use largely default settings, so their reported performance is a reasonable baseline rather than their achievable ceiling.
6. **Statistical scope.** The tests assess differences on these specific test sets and do not establish that the advantages would generalise.
